```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import requests
import pandas as pd
url = "https://api.coingecko.com/api/v3/coins/bitcoin/market_chart"
params = {
    "vs_currency": "usd",
    "days": "90"  # last 90 days of data
}
response = requests.get(url, params=params)
data = response.json()

# data['prices'] is a list of [timestamp, price] pairs
prices_df = pd.DataFrame(data['prices'], columns=['timestamp', 'price'])
prices_df['timestamp'] = pd.to_datetime(prices_df['timestamp'], unit='ms')
prices_df.head()
```

```python
plt.figure(figsize=(10,5))
plt.plot(prices_df['timestamp'], prices_df['price'])
plt.title("Bitcoin Price - Last 90 Days")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.show()
```

```python
import requests
import pandas as pd
url = "https://api.coingecko.com/api/v3/coins/bitcoin/ohlc"
params = {
    "vs_currency": "usd",
    "days": "90"
}
response = requests.get(url, params=params)
ohlc_data = response.json()

ohlc_df = pd.DataFrame(ohlc_data, columns=['timestamp', 'open', 'high', 'low', 'close'])
ohlc_df['timestamp'] = pd.to_datetime(ohlc_df['timestamp'], unit='ms')
ohlc_df.head(10)
```

```python
ohlc_df['daily_range'] = ohlc_df['high'] - ohlc_df['low']
ohlc_df[['timestamp', 'high', 'low', 'daily_range']].head(10)
```

```python
# Single column (returns a Series)
closes = ohlc_df['close']

# Multiple columns (returns a DataFrame)
subset = ohlc_df[['timestamp', 'close', 'daily_range']]
subset.head()
```

```python
# Find days where daily_range was unusually large
avg_range = ohlc_df['daily_range'].mean()
volatile_days = ohlc_df[ohlc_df['daily_range'] > avg_range]
volatile_days
```

```python
# Last 30 days only
last_30 = ohlc_df[ohlc_df['timestamp'] >= ohlc_df['timestamp'].max() - pd.Timedelta(days=30)]
last_30.head()
```

```python
print("Mean close price:", ohlc_df['close'].mean())
print("Std dev of close price:", ohlc_df['close'].std())
print("Min close:", ohlc_df['close'].min())
print("Max close:", ohlc_df['close'].max())
print("Mean daily range:", ohlc_df['daily_range'].mean())
```

```python
# Top 5 most volatile days
ohlc_df.sort_values('daily_range', ascending=False).head(5)
```

```python
ohlc_df['return'] = ohlc_df['close'].pct_change()
ohlc_df[['timestamp', 'close', 'return']].head(10)
```

```python
window = 7  # 7-day rolling window
ohlc_df['rolling_volatility'] = ohlc_df['return'].rolling(window=window).std()
ohlc_df[['timestamp', 'return', 'rolling_volatility']].head(15)
```

```python
plt.figure(figsize=(12,5))
plt.plot(ohlc_df['timestamp'], ohlc_df['rolling_volatility'])
plt.title("7-Day Rolling Volatility - Bitcoin")
plt.xlabel("Date")
plt.ylabel("Volatility (std dev of returns)")
plt.show()
```

```python
ohlc_df['rolling_volatility_3d'] = ohlc_df['return'].rolling(window=3).std()
ohlc_df['rolling_volatility_14d'] = ohlc_df['return'].rolling(window=14).std()
```

```python
fig, ax1 = plt.subplots(figsize=(12,6))

# Price on left axis
ax1.plot(ohlc_df['timestamp'], ohlc_df['close'], color='blue', label='Close Price')
ax1.set_xlabel('Date')
ax1.set_ylabel('Price (USD)', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Volatility on right axis
ax2 = ax1.twinx()
ax2.plot(ohlc_df['timestamp'], ohlc_df['rolling_volatility'], color='red', label='7-Day Rolling Volatility')
ax2.set_ylabel('Volatility', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title("Bitcoin Price vs Rolling Volatility")
fig.tight_layout()
plt.show()
```

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# 1. Fetch data
url = "https://api.coingecko.com/api/v3/coins/bitcoin/ohlc"
params = {"vs_currency": "usd", "days": "90"}
response = requests.get(url, params=params)
ohlc_df = pd.DataFrame(response.json(), columns=['timestamp', 'open', 'high', 'low', 'close'])
ohlc_df['timestamp'] = pd.to_datetime(ohlc_df['timestamp'], unit='ms')

# 2. Feature engineering
ohlc_df['daily_range'] = ohlc_df['high'] - ohlc_df['low']
ohlc_df['return'] = ohlc_df['close'].pct_change()
ohlc_df['rolling_volatility'] = ohlc_df['return'].rolling(window=7).std()

# 3. Quick sanity check
print(ohlc_df.tail())
```
