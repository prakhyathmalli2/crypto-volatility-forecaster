```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
image = cv2.imread("cat.jpg")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
# Create an initial mask
mask = np.zeros(image.shape[:2], np.uint8)
# Models for GrabCut
bgModel = np.zeros((1,65), np.float64)
fgModel = np.zeros((1,65), np.float64)
rect = (50, 20, 300, 450)
cv2.grabCut(image, mask, rect, bgModel, fgModel, 5, cv2.GC_INIT_WITH_RECT)
# Create binary mask: 1 for foreground, 0 for background
matting_mask = np.where((mask == 2) | (mask == 0), 0, 1).astype("uint8")
# Apply mask on image
result = image_rgb * matting_mask[:, :, np.newaxis]
plt.figure(figsize=(14,6))
plt.subplot(1,3,1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1,3,2)
plt.imshow(matting_mask, cmap="gray")
plt.title("Matting Mask")
plt.axis("off")
plt.subplot(1,3,3)
plt.imshow(result)
plt.title("Foreground Extracted")
plt.axis("off")
plt.show()

```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
img = cv2.imread('cat.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(16, 12))
plt.suptitle("Feature Detection & Extraction Practice", fontsize=18)
# -------------------------------------------------------------
# 1. EDGE DETECTION (Using Canny)
# -------------------------------------------------------------
edges = cv2.Canny(gray, 100, 200)
plt.subplot(2, 3, 1)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection (Canny)")
plt.axis("off")


```

```python
# -------------------------------------------------------------
# 2. CORNER DETECTION (Harris Corner)
# -------------------------------------------------------------
gray32 = np.float32(gray)
harris = cv2.cornerHarris(gray32, 2, 3, 0.04)
harris = cv2.dilate(harris, None)
img_harris = img.copy()
img_harris[harris > 0.01 * harris.max()] = [255, 0, 0]
plt.subplot(2, 3, 2)
plt.imshow(cv2.cvtColor(img_harris, cv2.COLOR_BGR2RGB))
plt.title("Corner Detection (Harris)")
plt.axis("off")
```

```python
# -------------------------------------------------------------
# 3. BLOB DETECTION (SimpleBlobDetector)
# -------------------------------------------------------------
params = cv2.SimpleBlobDetector_Params()
params.filterByArea = True
params.minArea = 100
detector = cv2.SimpleBlobDetector_create(params)
keypoints_blobs = detector.detect(gray)
img_blobs = cv2.drawKeypoints(img, keypoints_blobs, None, (0, 255, 0),                         cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
plt.subplot(2, 3, 3)
plt.imshow(cv2.cvtColor(img_blobs, cv2.COLOR_BGR2RGB))
plt.title("Blob Detection")
plt.axis("off")

```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from ultralytics import YOLO

# ----------------------------------------
# Load YOLO
# ----------------------------------------
model = YOLO("yolo11n.pt")

# ----------------------------------------
# Read classroom image
# ----------------------------------------
img = cv2.imread("classroom1.png")

if img is None:
    raise FileNotFoundError("classroom1.png not found")

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# ----------------------------------------
# YOLO Object Detection
# ----------------------------------------
results = model(img)

# Get image with bounding boxes
detected_img = results[0].plot()

# Convert BGR -> RGB for matplotlib
detected_rgb = cv2.cvtColor(detected_img, cv2.COLOR_BGR2RGB)

# ----------------------------------------
# Display
# ----------------------------------------
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Classroom")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(detected_rgb)
plt.title("YOLO Object Detection")
plt.axis("off")

plt.tight_layout()
plt.show()

print("Number of objects detected:", len(results[0].boxes))
```

```python
import cv2
import numpy as np

# Use 0 for webcam, or put a filename like "campus.mp4"
VIDEO_SOURCE = 0

cap = cv2.VideoCapture(VIDEO_SOURCE)

# Check camera/video
if not cap.isOpened():
    raise RuntimeError("Could not open webcam/video source.")

# Background subtractor for motion detection
bg = cv2.createBackgroundSubtractorMOG2(
    history=300,
    varThreshold=25,
    detectShadows=True
)

# Simple centroid tracker
next_id = 1
tracks = {}  # id -> (cx, cy)

max_objects = 0
motion_frames = 0


def dist(a, b):
    return np.sqrt(
        (a[0] - b[0]) ** 2 +
        (a[1] - b[1]) ** 2
    )


while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Resize frame
    frame = cv2.resize(frame, (960, 540))

    # -----------------------------------
    # 1. Background subtraction
    # -----------------------------------
    fg = bg.apply(frame)

    # -----------------------------------
    # 2. Clean foreground mask
    # -----------------------------------
    fg = cv2.GaussianBlur(fg, (5, 5), 0)

    _, fg = cv2.threshold(
        fg,
        200,
        255,
        cv2.THRESH_BINARY
    )

    kernel = np.ones((3, 3), np.uint8)

    fg = cv2.morphologyEx(
        fg,
        cv2.MORPH_OPEN,
        kernel,
        iterations=2
    )

    fg = cv2.dilate(
        fg,
        kernel,
        iterations=2
    )

    # -----------------------------------
    # 3. Find moving objects
    # -----------------------------------
    contours, _ = cv2.findContours(
        fg,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    detections = []

    for c in contours:

        area = cv2.contourArea(c)

        # Ignore small noisy regions
        if area < 800:
            continue

        x, y, w, h = cv2.boundingRect(c)

        cx = x + w // 2
        cy = y + h // 2

        detections.append(
            (x, y, w, h, cx, cy)
        )

    # -----------------------------------
    # 4. Count frames containing motion
    # -----------------------------------
    if len(detections) > 0:
        motion_frames += 1

    # -----------------------------------
    # 5. Simple centroid tracking
    # -----------------------------------
    assigned = set()
    new_tracks = {}

    for (x, y, w, h, cx, cy) in detections:

        best_id = None
        best_d = 50

        # Find nearest existing tracked object
        for tid, (tx, ty) in tracks.items():

            d = dist(
                (cx, cy),
                (tx, ty)
            )

            if d < best_d and tid not in assigned:

                best_d = d
                best_id = tid

        # If no nearby existing track,
        # create a new object ID
        if best_id is None:

            best_id = next_id
            next_id += 1

        assigned.add(best_id)

        new_tracks[best_id] = (
            cx,
            cy
        )

        # -----------------------------------
        # Draw bounding box
        # -----------------------------------
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Draw object ID
        cv2.putText(
            frame,
            f"ID {best_id}",
            (x, max(y - 8, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

        # Draw centroid
        cv2.circle(
            frame,
            (cx, cy),
            4,
            (0, 0, 255),
            -1
        )

    # Replace old tracks
    tracks = new_tracks

    # Maximum number of objects
    max_objects = max(
        max_objects,
        len(tracks)
    )

    # -----------------------------------
    # 6. Display information
    # -----------------------------------
    cv2.putText(
        frame,
        f"Moving objects: {len(tracks)}",
        (20, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 255),
        2
    )

    cv2.putText(
        frame,
        f"Motion frames: {motion_frames}",
        (20, 60),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 255),
        2
    )

    cv2.putText(
        frame,
        f"Unique IDs: {next_id - 1}",
        (20, 90),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (255, 255, 255),
        2
    )

    # -----------------------------------
    # 7. Show windows
    # -----------------------------------
    cv2.imshow(
        "Video Analysis: Motion + Tracking",
        frame
    )

    cv2.imshow(
        "Foreground Mask",
        fg
    )

    # Press q to stop
    key = cv2.waitKey(1) & 0xFF

    if key == ord("q"):
        break


# -----------------------------------
# Cleanup
# -----------------------------------
cap.release()
cv2.destroyAllWindows()


# -----------------------------------
# Summary
# -----------------------------------
print("\n=== Activity Summary ===")
print("Total motion frames:", motion_frames)
print("Max moving objects at once:", max_objects)
print("Total unique IDs created:", next_id - 1)
```
